```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./data/diabete_em_indigenas.csv')
```

```python
# verificando a distribuição dos valores nas colunas
columns = ['num_gestacoes', 'glicose', 'pressao_diastolica',
            'espessura_dobra_triceps', 'insulina', 'imc', 'predisposicao_diabetes',
            'idade']

fig, axes = plt.subplots(4,2, figsize=(8,8))
axes = axes.flatten()

for ax, cols in zip(axes, columns):
    ax.hist(x=df[cols], bins=10)
    ax.set_title(cols)
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
# observando a correlação das colunas com o alvo
plt.figure(figsize=(8,5))
sns.heatmap(df.corr()[['resultado']].sort_values(by='resultado', ascending=False), annot=True, vmax=1, vmin=-1, center=0)
plt.title('Correlação das variáveis com o alvo')
plt.show()
```

```python
# relação entre as duas variáveis mais forte e a variável alvo
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,
    x='imc',
    y='glicose',
    hue='resultado',
    palette={0: 'steelblue', 1: 'crimson'},
    alpha=0.75
)
plt.title('imc e glicose por diagnóstico de diabetes')
plt.legend(title='Resultado', labels=['Sem diabetes', 'Com diabetes'])
plt.grid(True)
plt.show()
```

```python
print("Verificando como os dados estão distribuídos")

# Buscando os dados com valor True e False, depois contando a quantidade de ocorrência de cada um
num_true = len(df.loc[df['resultado'] == 1])
num_false = len(df.loc[df['resultado'] == 0])

total = len(df.index)

print(f"Número de Casos Verdadeiros: {num_true} ({(num_true/total) * 100:.2f}%)")
print(f"Número de Casos Falsos: {num_false} ({(num_false/ total) * 100:.2f}%)")
```
