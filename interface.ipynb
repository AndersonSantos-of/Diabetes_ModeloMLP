```python
import gradio as gr
import pandas as pd
import joblib
```

```python
model = joblib.load('./model/model_MLP.pkl')
preprocessor = joblib.load('./model/preprocessor.pkl')
```

```python
def classification(arquivo):

    df = pd.read_csv(arquivo.name)
    df_transformed = preprocessor.fit_transform(df)

    y_pred = model.predict(df_transformed)

    df['diabete_pred'] = y_pred
    df.to_csv('./DataClassification.csv', index=False)

    return './DataClassification.csv'
```

```python
app = gr.Interface(
    classification,
    gr.File(file_types=['.csv']),
    'file'
)

app.launch(share=True)
```
