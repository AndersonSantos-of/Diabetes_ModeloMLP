### LINK DO DATASET NO KAGGLE

https://www.kaggle.com/datasets/pushprajojha/dataset

`num_preg → num_gestacoes`. Número de vezes que a mulher já engravidou.

`glucose_conc → glicose`. Concentração de açúcar no sangue depois de um teste de tolerância à glicose.

`diastolic_bp → pressao_diastolica`. É o segundo número da pressão arterial, o menor, medido em mmHg (o famoso "12 por 8": o 8 é a diastólica).

`thickness → espessura_dobra_triceps`. Espessura da dobra de pele medida no braço, na parte de trás (tríceps), em milímetros.

`insulin → insulina`. Nível de insulina no sangue duas horas depois do teste de glicose.

`bmi → imc`. Índice de massa corporal, que relaciona peso e altura. Em português a sigla é IMC, não BMI.

`diab_pred → predisposicao_diabetes`. É a função de histórico familiar de diabetes, um número que resume o quanto a família da pessoa tem casos da doença.

`age → idade`. Idade.

`skin → pele_padronizada`. É a mesma coisa que "thickness" (espessura da dobra de pele), só que em outra escala: cada valor dela é exatamente o valor de "thickness" dividido por 25,38, para todas as linhas. Ou seja, ela não traz nenhuma informação nova e repete a coluna anterior.

```python
import pandas as pd
```

```python
df = pd.read_csv('./data/pima-data.csv')
df.head()
```

```python
# traduzindo as colunas
traducao_colunas = {
    'num_preg':'num_gestacoes',
    'glucose_conc':'glicose',
    'diastolic_bp':'pressao_diastolica',
    'thickness':'espessura_dobra_triceps',
    'insulin':'insulina',
    'bmi':'imc',
    'diab_pred':'predisposicao_diabetes',
    'age':'idade',
    'diabetes':'resultado'
}

df.rename(columns = traducao_colunas, inplace=True)

# removendo a coluna com informação duplicada
df = df.drop(columns='skin')
```

```python
df.info()
```

```python
df.describe()
```

```python
# consultando quantos valores 0 tem em cada coluna
colunas = ['num_gestacoes', 'glicose', 'pressao_diastolica',
            'espessura_dobra_triceps', 'insulina', 'imc', 'predisposicao_diabetes',
            'idade']

quantidade = {}

for col in colunas:
    n = len(df.loc[df[col] == 0])
    quantidade.update({col:n})

quantidade
```

```python
# criando uma coluna extra para indicar onde foi colocado o valor da mediana
df["insulina_medida"] = df["insulina"] == 0

# calcula a mediana usando só os valores diferentes de zero, depois substitui no dataframe
mediana = df.loc[df["insulina"] != 0, "insulina"].median()
df["insulina"] = df["insulina"].replace(0, mediana)

print(mediana)
print(df["insulina_medida"].value_counts())
```

```python
# criando uma coluna extra para indicar onde foi colocado o valor da mediana
df['espessura_dobra_triceps_mediana'] = df['espessura_dobra_triceps'] == 0

# calcula a mediana usando só os valores diferentes de zero, depois substitui no dataframe
mediana = df.loc[df['espessura_dobra_triceps'] != 0, 'espessura_dobra_triceps'].median()
df['espessura_dobra_triceps'] = df['espessura_dobra_triceps'].replace(0, mediana)

print(mediana)
print(df["espessura_dobra_triceps_mediana"].value_counts())
```

```python
colunas = [
    "pressao_diastolica",
    "glicose",
    "imc",
]

# Troca cada zero pela média dos valores que não são zero
for coluna in colunas:
    media = df.loc[df[coluna] != 0, coluna].mean()
    df[coluna] = df[coluna].replace(0, media)
```

```python
# transformando os valores booleno em númericos
numbool = {
    True:1,
    False:0
}

df['espessura_dobra_triceps_mediana'] = df['espessura_dobra_triceps_mediana'].replace(numbool)
df['insulina_medida'] = df['insulina_medida'].replace(numbool)
df['resultado'] = df['resultado'].replace(numbool)
```
